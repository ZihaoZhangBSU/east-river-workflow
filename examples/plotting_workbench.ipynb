# East River plotting workbench

Current v0.8.0 workbench. `prepare_analysis_products` preprocesses the single hourly-internal-Time Noah-MP LDASOUT file once and writes reusable daily/cached products. Plotting then reuses those products; it does not read LDASIN forcing files.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from east_river_workflow.config import load_config
from east_river_workflow.plotting import configure_matplotlib, generate_all_figures, plot_watershed_comparison, plot_station_comparison
from east_river_workflow.workflow import prepare_analysis_products

CONFIG_CANDIDATES = [Path("../config/east_river_config.yaml"), Path("config/east_river_config.yaml")]
CONFIG_PATH = next(path.resolve() for path in CONFIG_CANDIDATES if path.exists())
cfg = load_config(CONFIG_PATH)
cfg.data["plotting"]["close_after_save"] = False
configure_matplotlib(cfg)
```

## Prepare the complete validated product set once

```python
products = prepare_analysis_products(cfg, run_validation=True)
```

## Generate all 26 baseline + 15 scientific figures

```python
figure_paths = generate_all_figures(cfg, products)
len(figure_paths)
```

## Replot the watershed SWE figure without rereading raw inputs

```python
plot_watershed_comparison(cfg, products["isnobal_watershed"], products["noah_watershed"], "SWE")
plt.show()
```

## Replot one direct SNOTEL SWE comparison

The station figure contains SNOTEL, the nearest iSnobal pixel, the nearest Noah-MP pixel, and the iSnobal min–max envelope from all valid iSnobal pixel centers inside that nearest Noah-MP cell. The same function supports snow depth.

```python
station = cfg.section("snotel")["stations"][0]
plot_station_comparison(cfg, int(station["id"]), station["name"], products["snotel"], products["isnobal_station"], products["noah_station"], "SWE")
plt.show()
```
